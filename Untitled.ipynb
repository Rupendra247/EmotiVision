```python
import cv2          # pip install opencv-python
# pip install opencv-contrib-python   (full package, needed for some cascades)
from deepface import DeepFace   # pip install deepface

# Load the Haar cascade for face detection
faceCascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

# Try camera index 1 first, fall back to 0
cap = cv2.VideoCapture(1)
if not cap.isOpened():
    cap = cv2.VideoCapture(0)
if not cap.isOpened():
    raise IOError("Cannot open webcam")

font = cv2.FONT_HERSHEY_SIMPLEX

while True:
    ret, frame = cap.read()
    if not ret:
        print("Failed to grab frame")
        break

    # --- Emotion detection ---
    try:
        result = DeepFace.analyze(frame, actions=['emotion'], enforce_detection=False)
        # DeepFace returns a list of dicts (one per detected face) in recent versions
        if isinstance(result, list):
            dominant_emotion = result[0]['dominant_emotion']
        else:
            dominant_emotion = result['dominant_emotion']
    except Exception as e:
        dominant_emotion = "No face"
        print("DeepFace error:", e)

    # --- Face rectangle detection ---
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = faceCascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=4, minSize=(30, 30))

    for (x, y, w, h) in faces:
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

    # --- Overlay emotion text ---
    cv2.putText(frame,
                dominant_emotion,
                (50, 50),
                font, 1.5,
                (0, 0, 255),
                2,
                cv2.LINE_AA)

    cv2.imshow('Original video', frame)

    if cv2.waitKey(2) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2          # pip install opencv-python
# pip install opencv-contrib-python   (full package, needed for some cascades)
from deepface import DeepFace   # pip install deepface

# Load the Haar cascade for face detection
faceCascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

# Try camera index 1 first, fall back to 0
cap = cv2.VideoCapture(1)
if not cap.isOpened():
    cap = cv2.VideoCapture(0)
if not cap.isOpened():
    raise IOError("Cannot open webcam")

font = cv2.FONT_HERSHEY_SIMPLEX

frame_count = 0
ANALYZE_EVERY_N_FRAMES = 10
dominant_emotion = "Detecting..."

while True:
    ret, frame = cap.read()
    if not ret:
        print("Failed to grab frame")
        break

    frame_count += 1

    # --- Face rectangle detection (runs every frame, it's cheap) ---
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    faces = faceCascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=4, minSize=(30, 30))

    for (x, y, w, h) in faces:
        cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

        # --- Emotion detection (only every N frames, and only on the cropped face) ---
        if frame_count % ANALYZE_EVERY_N_FRAMES == 0:
            face_roi = frame[y:y + h, x:x + w]
            try:
                result = DeepFace.analyze(
                    face_roi,
                    actions=['emotion'],
                    enforce_detection=False,
                    detector_backend='skip'  # skip re-detection, we already cropped
                )
                dominant_emotion = result[0]['dominant_emotion'] if isinstance(result, list) else result['dominant_emotion']
            except Exception as e:
                dominant_emotion = "No face"
                print("DeepFace error:", e)

        # Only need to analyze the first detected face per frame
        break

    # --- Overlay emotion text ---
    cv2.putText(frame,
                dominant_emotion,
                (50, 50),
                font, 1.5,
                (0, 0, 255),
                2,
                cv2.LINE_AA)

    cv2.imshow('Original video', frame)

    if cv2.waitKey(2) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
